### 변수 이름 규칙 

- 숫자로 시작 불가 : 1_name (X)
- 공백 불가 : user name (X)
- 하이픈(-) 불가 : user-age (X)
- 예약어 사용불가: if, def (X)


### 컨벤션

- 스네이크 표기법 : user_name, value_list

- 파스칼 케이스: UserName, ValueList
- 카멜 케이스 (단봉낙타법): userName
